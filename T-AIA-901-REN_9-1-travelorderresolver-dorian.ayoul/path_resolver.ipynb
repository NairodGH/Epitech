```python
# load data in pandas dataframe
# data are in the folder ./students_project/data_sncf/
agency = pd.read_csv('./students_project/data_sncf/agency.txt', sep=',')
calendar_dates = pd.read_csv('./students_project/data_sncf/calendar_dates.txt', sep=',')
calendar = pd.read_csv('./students_project/data_sncf/calendar.txt', sep=',')
routes = pd.read_csv('./students_project/data_sncf/routes.txt', sep=',')
stop_times = pd.read_csv('./students_project/data_sncf/stop_times.txt', sep=',')
stops = pd.read_csv('./students_project/data_sncf/stops.txt', sep=',')
transfers = pd.read_csv('./students_project/data_sncf/transfers.txt', sep=',')
trips = pd.read_csv('./students_project/data_sncf/trips.txt', sep=',')

timetables = pd.read_csv('./students_project/timetables.csv', sep='	')
mapping = pd.read_csv('./data/town_station_association.csv', sep='	')
```

```python
# Dijkstra test
def load_graph():
    graph = {}
    for index, row in timetables.iterrows():
        town1 = row['trajet'].split(" - ")[0]
        town2 = row['trajet'].split(" - ")[-1]
        distance = row['duree']
        if town1 not in graph:
            graph[town1] = []
        if town2 not in graph:
            graph[town2] = []
        graph[town1].append((town2, distance))
        graph[town2].append((town1, distance))

    return graph

def dijkstra(graph, start, end):
    # Priority queue to store (cost, town) tuples
    queue = [(0, start)]
    # Dictionary to store the shortest distance to each town
    distances = {town: float('inf') for town in graph}
    distances[start] = 0
    # Dictionary to store the shortest path to each town
    previous_town = {town: None for town in graph}

    while queue:
        current_distance, current_town = heapq.heappop(queue)

        if current_town == end:
            break

        for neighbor, weight in graph[current_town]:
            distance = current_distance + weight

            if distance < distances[neighbor]:
                distances[neighbor] = distance
                previous_town[neighbor] = current_town
                heapq.heappush(queue, (distance, neighbor))

    # Reconstruct the shortest path
    path = []
    current_town = end
    while previous_town[current_town] is not None:
        path.insert(0, current_town)
        current_town = previous_town[current_town]
    if path:
        path.insert(0, start)

    return path, distances[end] if distances[end] != float('inf') else None


graph = load_graph()
print(dijkstra(graph, "Gare de Le Havre", "Gare de Trouville-Deauville"))
```

```python
input_test = "Toulouse,Bordeaux"

def get_station(input_test):
    town1, town2 = input_test.split(",")
    station1 = mapping[mapping['town'] == town1]['station'].values[0]
    station2 = mapping[mapping['town'] == town2]['station'].values[0]
    graph = load_graph()
    path, distance = dijkstra(graph, station1, station2)
    return path

print(get_station(input_test))
```
